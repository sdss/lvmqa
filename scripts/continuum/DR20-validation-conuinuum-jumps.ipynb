```python
from collections import defaultdict
import json
from pathlib import Path

def find_flagged_expnums(folder, threshold=10):
    folder = Path(folder)

    for json_file in folder.rglob("*.json"):
        try:
            with open(json_file, "r") as f:
                data = json.load(f)
        except Exception as e:
            print(f"Skipping {json_file}: {e}")
            continue

        expnum = data["expnum"]
        n_flagged_blue = data["extended_report"]["continuum_check_blue_red"]["n_flagged"]
        n_flagged_red = data["extended_report"]["continuum_check_red_ir"]["n_flagged"]

        if (n_flagged_blue >= threshold) or (n_flagged_red >= threshold):
            print(f"expnum={expnum}: jump flagged: blue={n_flagged_blue} red={n_flagged_red}")

folder = "./logs_v120/"
threshold = 10
find_flagged_expnums(folder, threshold)
```
