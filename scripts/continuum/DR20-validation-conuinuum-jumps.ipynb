```python
from collections import defaultdict
import json
from pathlib import Path

def find_flagged_expnums(folder):
    folder = Path(folder)

    n_warn = 0
    n_bad = 0
    for json_file in folder.rglob("*.json"):
        try:
            with open(json_file, "r") as f:
                data = json.load(f)
        except Exception as e:
            print(f"Skipping {json_file}: {e}")
            continue

        expnum = data["expnum"]
        n_flagged_blue = data["extended_report"]["continuum_check_blue_red"]["n_flagged_warn"]
        n_flagged_red = data["extended_report"]["continuum_check_red_ir"]["n_flagged_warn"]
        warn = data["continuum_jump_warning_flag"]
        bad = data["continuum_jump_bad_flag"]

        if warn or bad:
            n_warn += warn
            n_bad += bad
            print(f"expnum={expnum}: jump flagged: blue={n_flagged_blue} red={n_flagged_red} warn={warn} bad={bad}")

    print('n_warn =', n_warn)
    print('n_bad =', n_bad)


folder = "./logs_v120_jump_final/"
find_flagged_expnums(folder)
```

```python
from pathlib import Path
import json
import matplotlib.pyplot as plt
import numpy as np

def collect_jump_lists(folder, arms="continuum_check_blue_red"):
    folder = Path(folder)
    all_jumps = []

    for json_file in folder.rglob("*.json"):
        try:
            with open(json_file, "r") as f:
                data = json.load(f)
        except Exception as e:
            print(f"Skipping {json_file}: {e}")
            continue

        try:
            jumps = data["extended_report"][arms]["jump_list"]
        except KeyError:
            print(f"Missing jump_list in {json_file} for {arms}")
            continue

        if jumps is None:
            continue

        if isinstance(jumps, list):
            all_jumps.extend(jumps)
        else:
            all_jumps.append(jumps)

    # Keep only numeric values
    all_jumps = [x for x in all_jumps if isinstance(x, (int, float))]
    return all_jumps


folder = "./logs_v120_jump_final_jumplist/"

jumps_blue_red = collect_jump_lists(folder, arms="continuum_check_blue_red")
jumps_red_ir = collect_jump_lists(folder, arms="continuum_check_red_ir")

print(f"Collected {len(jumps_blue_red)} blue-red jump values")
print(f"Collected {len(jumps_red_ir)} red-ir jump values")

def print_stats(arr, name):
    arr = np.array(arr)

    arr = arr[np.isfinite(arr)]

    median = np.median(arr)
    p16, p50, p84 = np.percentile(arr, [16, 50, 84])

    print(f"\n{name}:")
    print(f"  N = {len(arr)}")
    print(f"  Median = {median:.3g}")
    print(f"  16th percentile = {p16:.3g}")
    print(f"  50th percentile = {p50:.3g}")
    print(f"  84th percentile = {p84:.3g}")

print_stats(jumps_blue_red, "b-r jumps")
print_stats(jumps_red_ir, "r-z jumps")


scale=0.75
fig, axes = plt.subplots(2, 1, figsize=(scale*8, scale*10), sharex=True)

axes[0].hist(jumps_blue_red, bins=100)
axes[0].set_yscale("log")
axes[0].set_ylabel("N")
axes[0].set_title("b-r jump")

axes[1].hist(jumps_red_ir, bins=100)
axes[1].set_yscale("log")
axes[1].set_xlabel("Jump value, (blue-red)/mad_std")
axes[1].set_ylabel("N")
axes[1].set_title("r-z jump")

plt.tight_layout()
plt.savefig('count_jump_hist_v2.png', dpi=150)
```
